**Augmented Dickey-Fuller (ADF) Test**


The Augmented Dickey-Fuller (ADF) Test is a statistical test used to determine whether a time series is stationary or non-stationary.

A stationary time series has statistical properties, such as the mean and variance, that remain relatively constant over time. Many forecasting models, including SARIMA, require stationary data for accurate predictions.

Hypotheses:

Null Hypothesis (H₀): The time series is non-stationary.
Alternative Hypothesis (H₁): The time series is stationary.

```python
from statsmodels.tsa.stattools import adfuller

# Perform ADF Test
adf_result = adfuller(monthly_sales_ts['Sales'])

print("ADF Statistic:", adf_result[0])
print("p-value:", adf_result[1])

print("\nCritical Values:")
for key, value in adf_result[4].items():
    print(f"{key}: {value}")
```

**ADF Test Interpretation**

The Augmented Dickey-Fuller (ADF) test produced an ADF statistic of -4.4161 and a p-value of 0.000278, which is less than the significance level of 0.05. Therefore, the null hypothesis is rejected, indicating that the monthly sales time series is stationary. This means that the statistical properties of the series, such as its mean and variance, remain relatively stable over time. Since the data is already stationary, additional differencing is not required before applying forecasting models such as SARIMA.

Since the ADF test confirms that the monthly sales series is already stationary (p-value < 0.05), first-order differencing was not applied, as it is unnecessary. The stationary series can be used directly for forecasting models in the next task.

**Model 1 – SARIMA (Statistical Model)**

```python
!pip install statsmodels
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
```

```python
monthly_sales.head()
```

```python
sarima_data = monthly_sales.copy()

sarima_data.set_index('Order Date', inplace=True)

sarima_data.head()
```

```python
train = sarima_data.iloc[:-3]

test = sarima_data.iloc[-3:]

print(train.shape)
print(test.shape)
```

```python
order=(1,1,1)

seasonal_order=(1,1,1,12)
```

```python
model = SARIMAX(
    train['Sales'],
    order=(1,1,1),
    seasonal_order=(1,1,1,12),
    enforce_stationarity=False,
    enforce_invertibility=False
)

# Fit Model
sarima_model = model.fit()

# Model Summary
print(sarima_model.summary())
```

```python
forecast = sarima_model.get_forecast(steps=3)

forecast_mean = forecast.predicted_mean

confidence = forecast.conf_int()

print(forecast_mean)
```

```python
forecast_df = pd.DataFrame({
    'Actual Sales': test['Sales'],
    'Forecast Sales': forecast_mean
})

forecast_df
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae = mean_absolute_error(test['Sales'], forecast_mean)
rmse = np.sqrt(mean_squared_error(test['Sales'], forecast_mean))
mape = np.mean(np.abs((test['Sales'] - forecast_mean) / test['Sales'])) * 100

print("MAE :", round(mae,2))
print("RMSE:", round(rmse,2))
print("MAPE:", round(mape,2))
```

```python
print(train.index)
print(test.index)
print(type(train.index))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

# Training Data
plt.plot(
    train.index,
    train['Sales'],
    label='Training Data',
    color='blue'
)

# Actual Sales
plt.plot(
    test.index,
    test['Sales'],
    label='Actual Sales',
    color='green',
    marker='o',
    linewidth=2
)

# Forecasted Sales
plt.plot(
    forecast_mean.index,
    forecast_mean,
    label='Forecasted Sales',
    color='red',
    marker='o',
    linewidth=2
)

# Confidence Interval
plt.fill_between(
    confidence.index,
    confidence.iloc[:, 0],
    confidence.iloc[:, 1],
    color='pink',
    alpha=0.3,
    label='95% Confidence Interval'
)

plt.title("SARIMA Forecast vs Actual Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```

**Observation**

The SARIMA model generated forecasts for the next three months and was able to capture the general sales pattern. Although the predicted values differ from the actual sales, the model follows the overall trend and seasonal behavior of the data. The model achieved an MAE of 19,244.49, RMSE of 19,950.07, and MAPE of 20.53%. These metrics will be compared with Prophet and XGBoost to determine the most suitable forecasting model for production use.

**Model 3 – XGBoost for Time Series**

```python
!pip install xgboost
```

```python
import xgboost as xgb
import numpy as np
import pandas as pd

from sklearn.metrics import mean_absolute_error, mean_squared_error
import matplotlib.pyplot as plt
```

```python
xgb_data = monthly_sales.copy()

xgb_data.head()
```

```python
xgb_data['Lag_1'] = xgb_data['Sales'].shift(1)

xgb_data['Lag_2'] = xgb_data['Sales'].shift(2)

xgb_data['Lag_3'] = xgb_data['Sales'].shift(3)

xgb_data['Rolling_Mean_3'] = (
    xgb_data['Sales']
    .rolling(window=3)
    .mean()
)
```

```python
xgb_data['Month'] = xgb_data['Order Date'].dt.month

xgb_data['Quarter'] = xgb_data['Order Date'].dt.quarter
```

```python
def get_season(month):

    if month in [12,1,2]:
        return 1      # Winter

    elif month in [3,4,5]:
        return 2      # Spring

    elif month in [6,7,8]:
        return 3      # Summer

    else:
        return 4      # Autumn

xgb_data['Season'] = xgb_data['Month'].apply(get_season)
```

```python
xgb_data = xgb_data.dropna()

xgb_data.head()
```

```python
features = [
    'Lag_1',
    'Lag_2',
    'Lag_3',
    'Rolling_Mean_3',
    'Month',
    'Quarter',
    'Season'
]

X = xgb_data[features]

y = xgb_data['Sales']
```

```python
X_train = X.iloc[:-3]

X_test = X.iloc[-3:]

y_train = y.iloc[:-3]

y_test = y.iloc[-3:]
```

```python
model = xgb.XGBRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)

predictions
```

```python
mae = mean_absolute_error(y_test, predictions)

rmse = np.sqrt(mean_squared_error(y_test, predictions))

mape = np.mean(
    np.abs((y_test - predictions) / y_test)
) * 100

print("MAE :", round(mae,2))
print("RMSE:", round(rmse,2))
print("MAPE:", round(mape,2))
```

```python
forecast_df = pd.DataFrame({
    'Actual Sales': y_test.values,
    'Predicted Sales': predictions
},
index=xgb_data.iloc[-3:]['Order Date'])

forecast_df
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    xgb_data['Order Date'].iloc[:-3],
    y_train,
    label='Training',
    color='blue'
)

plt.plot(
    xgb_data['Order Date'].iloc[-3:],
    y_test,
    marker='o',
    linewidth=2,
    label='Actual',
    color='green'
)

plt.plot(
    xgb_data['Order Date'].iloc[-3:],
    predictions,
    marker='o',
    linewidth=2,
    label='Predicted',
    color='red'
)

plt.title("XGBoost Forecast vs Actual Sales")

plt.xlabel("Date")

plt.ylabel("Sales")

plt.legend()

plt.grid(True)

plt.show()
```

**Observation**


The XGBoost model successfully transformed the time series into a supervised learning problem using lag features, rolling averages, and calendar features. It generated accurate forecasts for the final three months of the dataset. Compared with SARIMA and Prophet, XGBoost achieved the lowest forecasting errors, demonstrating its ability to capture nonlinear relationships in the sales data.

 **Z-Score Based Anomaly Detection**

```python
weekly_sales['Rolling_Mean'] = weekly_sales['Sales'].rolling(window=8).mean()
weekly_sales['Rolling_STD'] = weekly_sales['Sales'].rolling(window=8).std()
```

```python
weekly_sales['Z_Score'] = (
    (weekly_sales['Sales'] - weekly_sales['Rolling_Mean'])
    / weekly_sales['Rolling_STD']
)
```

```python
weekly_sales['Z_Anomaly'] = weekly_sales['Z_Score'].abs() > 2

z_anomalies = weekly_sales[weekly_sales['Z_Anomaly']]

z_anomalies
```

**Task 6 — Product Demand Segmentation using Clustering**


```python
# Prepare the Data

cluster_df = df.copy()

cluster_df['Order Date'] = pd.to_datetime(cluster_df['Order Date'])

cluster_df['Year'] = cluster_df['Order Date'].dt.year

cluster_df.head()
```

```python
# Calculate Total Sales Volume

total_sales = (
    cluster_df.groupby('Sub-Category')['Sales']
              .sum()
              .rename('Total Sales')
)

total_sales
```

```python
# Calculate Year-over-Year Sales Growth

yearly_sales = (
    cluster_df.groupby(['Sub-Category', 'Year'])['Sales']
              .sum()
              .reset_index()
)

yearly_sales.head()
```

```python
growth = yearly_sales.pivot(
    index='Sub-Category',
    columns='Year',
    values='Sales'
)

growth
```

```python
growth_rate = (
    (growth[2018] - growth[2017])
    / growth[2017]
) * 100

growth_rate = growth_rate.rename('Growth Rate')

growth_rate
```

```python
# Calculate Sales Volatility

monthly_sales = (
    cluster_df.groupby(['Sub-Category', pd.Grouper(key='Order Date', freq='ME')])['Sales']
              .sum()
              .reset_index()
)

monthly_sales.head()
```

```python
# Calculate Volatility

volatility = (
    monthly_sales.groupby('Sub-Category')['Sales']
                 .std()
                 .rename('Volatility')
)

volatility
```

```python
# Calculate Average Order Value

average_order = (
    cluster_df.groupby('Sub-Category')['Sales']
              .mean()
              .rename('Average Order Value')
)

average_order
```

```python
# Combine All Features

features = pd.concat(
    [total_sales,
     growth_rate,
     volatility,
     average_order],
    axis=1
)

features
```

```python
# Standardize the Features

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_features = scaler.fit_transform(features)

scaled_features
```

```python
# Elbow Method

from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
inertia = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(scaled_features)
    inertia.append(kmeans.inertia_)
```

```python
inertia = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(scaled_features)
    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))
plt.plot(range(1,11), inertia, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.grid(True)
plt.show()
```

```python
# Apply K-Means Clustering

# Apply K-Means with 4 clusters

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

features['Cluster'] = kmeans.fit_predict(scaled_features)

features
```
